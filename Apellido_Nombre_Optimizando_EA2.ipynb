## 5. Gestión de Versiones con Git y GitHub (Ramas, Commits y Pull Requests)

A continuación se inspecciona directamente desde el entorno de ejecución el estado del repositorio Git, las ramas creadas bajo el modelo **GitFlow** y el grafo histórico de *commits* y fusiones de *Pull Requests*.


```python
import subprocess

print("=== 1. LISTADO DE RAMAS DEL REPOSITORIO (GITFLOW) ===")
branches = subprocess.check_output(["git", "branch", "-a", "-v"], text=True, encoding="utf-8")
print(branches)

print("=== 2. GRAFO DE COMMITS, FUSIONES DE RAMAS Y PULL REQUESTS ===")
git_graph = subprocess.check_output(["git", "log", "--graph", "--oneline", "--decorate", "--all"], text=True, encoding="utf-8")
print(git_graph)
```

## 6. Integración Continua (CI): Ejecución de Pruebas Automatizadas

El archivo `tests/test_scraper.py` implementa **6 pruebas automatizadas** con `unittest` que actúan como compuerta de calidad (*Quality Gate*) dentro del pipeline de GitHub Actions:
* `test_01_html_extraction_count`: Verifica la extracción exacta de tarjetas de producto desde un DOM simulado de Mercado Libre.
* `test_02_price_cleaning_and_numeric_type`: Comprueba la limpieza de puntos separadores de miles y el tipado entero (`int`) en pesos colombianos.
* `test_03_storage_capacity_normalization`: Evalúa la expresión regular que clasifica las capacidades (`256GB`, `512GB`, `1TB`).
* `test_04_shipping_and_url_sanitization`: Verifica la limpieza de parámetros de rastreo (`#`) en las URLs y la estandarización de envíos.
* `test_05_production_csv_dataset_integrity`: Audita que el dataset `iphone_17_pro_max_mercadolibre.csv` contenga los 48 registros sin valores nulos.
* `test_06_production_json_dataset_integrity`: Audita la estructura y validez del archivo `iphone_17_pro_max_mercadolibre.json`.


```python
# Ejecución en vivo de la suite de pruebas unitarias y de integridad de datos
import unittest
from tests.test_scraper import TestMercadoLibreScraperPipeline

suite = unittest.TestLoader().loadTestsFromTestCase(TestMercadoLibreScraperPipeline)
runner = unittest.TextTestRunner(verbosity=2)
result = runner.run(suite)

print(f"\n[RESUMEN CI] Total Pruebas: {result.testsRun} | Exitosas: {result.testsRun - len(result.failures) - len(result.errors)} | Fallidas: {len(result.failures)}")
```
